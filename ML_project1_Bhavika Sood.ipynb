# 1. Load the Dataset

```python
import pandas as pd

data = '/Users/bhavikasood/Downloads/Project_1_python/dataset_adult/adult.data'
test = '/Users/bhavikasood/Downloads/Project_1_python/dataset_adult/adult.test'

columns = ['age', 'workclass', 'fnlwgt', 'education', 'education_num', 'marital_status', 'occupation', 
           'relationship', 'race', 'sex', 'capital_gain', 'capital_loss', 'hours_per_week', 'native_country', 'income']

adult_data = pd.read_csv(data, names=columns, na_values='?', skipinitialspace=True)

adult_test = pd.read_csv(test, names=columns, na_values='?', skipinitialspace=True, skiprows=1)

print("\nAdult Training Data:\n")
display(adult_data.head())

print("\nAdult Test Data:\n")
display(adult_test.head())
```

# 2. Output the structure of the Dataset

```python
adult_datainfo = adult_data.info()
print('\n')
adult_testinfo = adult_test.info()
```

# 3. Clean the Dataset

```python
print("Missing Values in the Traing Data Before Cleaning")
print(adult_data.isnull().sum())

print("\nMissing Values in the Test Data Before Cleaning")
print(adult_test.isnull().sum())

clean_adult_data = adult_data.dropna()
clean_adult_test = adult_test.dropna()

clean_adult_test['income'] = clean_adult_test['income'].str.replace('.', '')

print("\nMissing Values in the Traing Data After Cleaning")
print(clean_adult_data.isnull().sum())

print("\nMissing Values in the Test Data After Cleaning")
print(clean_adult_test.isnull().sum())

print("\nAdult Training Data (after cleaning):\n")
display(clean_adult_data.head())
print("\nAdult Test Data (after cleaning):\n")
display(clean_adult_test.head())
```

```python
clean_adult_data['income'] = clean_adult_data['income'].apply(lambda x: 1 if x == '>50K' else 0)
clean_adult_test['income'] = clean_adult_test['income'].apply(lambda x: 1 if x == '>50K' else 0)

categorical_columns = ['workclass', 'education', 'marital_status', 'occupation', 
                       'relationship', 'race', 'sex', 'native_country']

adult_data_encoded = pd.get_dummies(clean_adult_data, columns=categorical_columns)
adult_test_encoded = pd.get_dummies(clean_adult_test, columns=categorical_columns)

adult_test_encoded = adult_test_encoded.reindex(columns=adult_data_encoded.columns, fill_value=0)

print("\nAdult Training Data (after encoding):\n")
display(adult_data_encoded.head())
print("\nAdult Test Data (after encoding):\n")
display(adult_test_encoded.head())
```

# 4. Explore the Data

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.histplot(clean_adult_data['age'], bins=30, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.countplot(data=clean_adult_data, x='sex', hue='income')
plt.title('Income Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Income', labels=['<=50K', '>50K'])

plt.tight_layout()
plt.show()
```

# 5. Predictive Modeling

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report

pmodel = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,   
    random_state=42
)

pmodel.fit(adult_data_encoded.drop('income', axis=1), adult_data_encoded['income'])

pred = pmodel.predict(adult_test_encoded.drop('income', axis=1))
```

#### Total no. of individuals who earn more than $50k a year

```python
totalnum = sum(pred == 1)
print(f"Number of individuals who earn more than USD 50,000/year: {totalnum}")

above50k = adult_test_encoded.index[pred == 1]
individuals = clean_adult_test.loc[above50k]

print("\nIndividuals who earn more than USD 50,000/year:\n")
display(individuals)
```

#### Accuracy

```python
accuracy = accuracy_score(adult_test_encoded['income'], pred)

print(f"Accuracy: {accuracy * 100:.3f}%")
```
